```python
from fractions import Fraction

# Every possible world for two dice: (first die, second die)
worlds = []
for first in range(1, 7):
    for second in range(1, 7):
        worlds.append((first, second))  # (1,1), (1,2), ..., (6,6)
print("Number of possible worlds:", len(worlds))

# Count the worlds where the two dice add up to a given total
def probability_of_sum(total):
    favorable = 0
    for first, second in worlds: # Iterate through all possible worlds in the tuple (first die, second die)
        if first + second == total:
            favorable = favorable + 1
    return Fraction(favorable, len(worlds)) # Fraction() returns the probability of the sum as a fraction of favorable worlds over total worlds

print("P(sum = 12) =", probability_of_sum(12))
print("P(sum = 7)  =", probability_of_sum(7))

# P(sum = 12 | first die = 6): keep only the worlds where the first die is 6
worlds_given = []
for first, second in worlds: # Iterate through all possible worlds in the tuple (first die, second die)
    if first == 6:
        worlds_given.append((first, second))
print("Worlds left after learning the first die is 6:", len(worlds_given))

# Count how many of those given worlds have a sum of 12
favorable = 0
for first, second in worlds_given: # Iterate through all possible worlds where first die is 6
    if first + second == 12:
        favorable = favorable + 1

p_sum_12_given_first_6 = Fraction(favorable, len(worlds_given))
print("P(sum = 12 | first die = 6) =", p_sum_12_given_first_6) 


p_rain_and_cloud = 0.08
p_rain, p_cloud = 0.10, 0.40

print(f"If independent, P(R and C) would be {p_rain * p_cloud:.2f}")
print(f"The table says P(R and C) = {p_rain_and_cloud:.2f}") 
# Numbers copied from the table (probabilities of each group of days)
# R = rain in the afternoon, C = cloudy morning
p_rain_and_cloud = 0.08   # P(R and C): cell "cloudy morning AND rain"
p_rain_and_clear = 0.02   # P(R and clear): cell "clear morning AND rain"
p_cloud = 0.40            # P(C): row total, all cloudy mornings
p_clear = 0.60            # P(clear): row total, all clear mornings
p_rain = 0.10             # P(R): column total, all rainy afternoons

# "Given that" = divide by the size of the group we already know about
rain_given_cloud = p_rain_and_cloud / p_cloud   # P(R | C) = P(R and C) / P(C)
cloud_given_rain = p_rain_and_cloud / p_rain    # P(C | R) = P(R and C) / P(R)
rain_given_clear = p_rain_and_clear / p_clear   # P(R | clear) = P(R and clear) / P(clear)

print(f"P(R|C)     = {rain_given_cloud:.1%}   rain after a cloudy morning")
print(f"P(C|R)     = {cloud_given_rain:.1%}   cloudy morning before a rainy afternoon")
print(f"P(R|clear) = {rain_given_clear:.1%}   rain after a clear morning")

# Checks: the results must match the hand calculations
assert abs(rain_given_cloud - 0.20) < 1e-12
assert abs(cloud_given_rain - 0.80) < 1e-12 
prior_rain = 0.10                   # P(R)
cloudy_rate = 0.40                  # P(C)
cloudy_given_rain = 0.80            # P(C | R), from the storm study

both = cloudy_given_rain * prior_rain          # P(R and C)
rain_given_cloudy = both / cloudy_rate         # Bayes' theorem

print(f"P(R and C) = {both:.2f}")
print(f"P(R|C) = {rain_given_cloudy:.1%}")
assert abs(rain_given_cloudy - 0.20) < 1e-12 
def should_bring_umbrella(chance_of_rain, dry_cost=2, soaked_cost=4):
    cost_of_bringing = dry_cost * (1 - chance_of_rain)
    cost_of_leaving = soaked_cost * chance_of_rain
    return cost_of_bringing < cost_of_leaving

for p in (0.10, 0.20, 0.40):
    print(f"P(rain) = {p:.0%}: bring umbrella? {should_bring_umbrella(p)}")
assert should_bring_umbrella(0.20) is False
assert should_bring_umbrella(0.40) is True 



# INDIVIDUAL ACTIVITY: BAYES' THEOREM

# Probability of a cloudy morning
cloudy_rate = 0.40

# Probability of clouds given that it rains
cloudy_given_rain = 0.80

# Calculate the probability of rain given a cloudy morning
def rain_chance_given_cloudy(rain_rate):
    return (cloudy_given_rain * rain_rate) / cloudy_rate

# Test three different initial probabilities of rain
for rain_rate in (0.05, 0.10, 0.25):
    posterior = rain_chance_given_cloudy(rain_rate)

    print(
        f"Rain rate: {rain_rate:.0%}; "
        f"rain chance after cloudy morning: {posterior:.1%}; "
        f"bring umbrella: {should_bring_umbrella(posterior)}"
    )

# Verify the results
assert abs(rain_chance_given_cloudy(0.05) - 0.10) < 1e-12
assert abs(rain_chance_given_cloudy(0.10) - 0.20) < 1e-12
assert abs(rain_chance_given_cloudy(0.25) - 0.50) < 1e-12

assert should_bring_umbrella(0.10) is False
assert should_bring_umbrella(0.20) is False
assert should_bring_umbrella(0.50) is True

print("Posterior calculations and umbrella decisions passed.")
```
